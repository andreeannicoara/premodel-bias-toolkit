```python
import pandas as pd

df = pd.read_csv('data/diabetic_data.csv', na_values=['?'], low_memory=False)

print("SHAPE")
print(df.shape)

print("\nRACE")
print(df['race'].value_counts(dropna=False))

print("\nTOP 10 MISSING")
print(df.isnull().sum().sort_values(ascending=False).head(10))

print("\nREADMITTED")
print(df['readmitted'].value_counts(dropna=False))

print("\nGENDER")
print(df['gender'].value_counts(dropna=False))
```

```python
print("READMITTED BY RACE")
print(pd.crosstab(df['race'], df['readmitted'], normalize='index').round(3) * 100)

print("\nREADMITTED BY GENDER")
print(pd.crosstab(df['gender'], df['readmitted'], normalize='index').round(3) * 100)

print("\nWEIGHT MISSING BY RACE")
print(df.groupby('race')['weight'].apply(lambda x: x.isnull().mean()).round(3) * 100)
```
